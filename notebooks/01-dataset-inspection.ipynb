```python
import os
import nibabel as nib
import numpy as np
import pandas as pd

from pathlib import Path

DATASET_PATH = Path(
    "/kaggle/input/datasets/i212385nomanarif/2024-brats-glioma"
)

print("Dataset exists:", DATASET_PATH.exists())
```

```python
patient_dirs = sorted([
    p for p in DATASET_PATH.iterdir()
    if p.is_dir()
])

print("Total patients:", len(patient_dirs))

print("\nFirst 10 patients:")
for patient in patient_dirs[:10]:
    print(patient.name)
```

```python
patient_id = "BraTS-GLI-02547-101"

patient_path = DATASET_PATH / patient_id

files = sorted(os.listdir(patient_path))

for file in files:
    print(file)
```

```python
for file in files:

    path = patient_path / file
    img = nib.load(path)

    print(f"\nFile: {file}")
    print("Shape:", img.shape)
    print("Voxel spacing:", img.header.get_zooms())
    print("Data type:", img.get_data_dtype())
    print("Affine matrix:\n", img.affine)
```

```python
modalities = ["t1n", "t1c", "t2w", "t2f"]

images = {}

for modality in modalities:

    file_path = patient_path / f"{patient_id}-{modality}.nii"

    img = nib.load(file_path)

    images[modality] = img

for modality, img in images.items():

    print(
        modality,
        "Shape:", img.shape,
        "Spacing:", img.header.get_zooms()
    )
```

```python
shapes = [img.shape for img in images.values()]

assert all(shape == shapes[0] for shape in shapes)

print("All modalities have matching dimensions.")
```

```python
affines = [img.affine for img in images.values()]

assert all(
    np.allclose(affines[0], affine)
    for affine in affines[1:]
)

print("All modalities have matching affine matrices.")
```

```python
seg_path = patient_path / f"{patient_id}-seg.nii"

seg_img = nib.load(seg_path)

seg = np.asanyarray(seg_img.dataobj)

labels, counts = np.unique(
    seg,
    return_counts=True
)

for label, count in zip(labels, counts):

    print(f"Label {label}: {count:,} voxels")
```

```python
total_voxels = seg.size

for label, count in zip(labels, counts):

    percentage = (count / total_voxels) * 100

    print(
        f"Label {label}: "
        f"{percentage:.4f}%"
    )
```

```python
metadata = []

for file in files:

    img = nib.load(patient_path / file)

    metadata.append({
        "File": file,
        "Shape": img.shape,
        "Voxel Spacing": img.header.get_zooms(),
        "Data Type": str(img.get_data_dtype()),
        "File Size (MB)": round(
            (patient_path / file).stat().st_size / (1024**2),
            2
        )
    })

metadata_df = pd.DataFrame(metadata)

metadata_df
```
